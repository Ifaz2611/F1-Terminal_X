# Lap Time Analysis
Lap-by-lap statistics and stint comparisons.

```python
import pandas as pd

from f1_terminal.features import engineer_lap_features
from f1_terminal.io import load_session_data

# Load laps (offline fallback to synthetic)
try:
    laps = load_session_data(2023, 'Monza', 'R', source='auto')
except Exception as e:
    print(f"Fallback: {e}")
    laps = pd.DataFrame({
        'Driver': ['VER','VER','HAM','HAM'],
        'LapNumber': [1,2,1,2],
        'LapTime': pd.to_timedelta(['1:22.100','1:21.800','1:22.500','1:22.000']),
        'Sector1Time': pd.to_timedelta(['0:27.1','0:26.9','0:27.3','0:27.0']),
        'Sector2Time': pd.to_timedelta(['0:28.0','0:27.8','0:28.2','0:28.0']),
        'Sector3Time': pd.to_timedelta(['0:27.0','0:27.1','0:27.0','0:27.0'])
    })

class _Sess:
    pass
_Sess.laps = laps
features = engineer_lap_features(_Sess(), drivers=['VER','HAM'])
print(features.head())
print(features.describe(include='all'))
```
