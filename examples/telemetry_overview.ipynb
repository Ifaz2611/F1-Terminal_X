# Telemetry Overview
Explore speed, throttle, brake, and gear traces.
Works offline via `data/telemetry_sample.csv`.

```python
import matplotlib.pyplot as plt

from f1_terminal.io import load_session_data

# Offline demo — no network needed
df = load_session_data(2023, 'Monza', 'Q', source='csv')
print(df.head())

plt.figure()
plt.plot(df['Distance'], df['Speed'])
plt.xlabel('Distance (m)')
plt.ylabel('Speed (km/h)')
plt.title('Sample Lap — Speed Trace')
plt.show()
```

```python
plt.figure()
plt.plot(df['Distance'], df['Throttle'], label='Throttle')
plt.plot(df['Distance'], df['Brake']*100, label='Brake')
plt.legend()
plt.xlabel('Distance (m)')
plt.show()
print('Throttle/Brake overlay')
```
